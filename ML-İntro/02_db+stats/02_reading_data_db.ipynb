```python
path = 'data/superheroes.db'
conn = sq3.Connection(path)
```

```python
query = """
SELECT *
FROM superheroes
WHERE appearance_count IS NOT NULL
AND first_appearance IS NOT NULL
AND eye_color IS NOT NULL
ORDER BY appearance_count desc
"""
data = pd.read_sql(query, conn)
data
```
